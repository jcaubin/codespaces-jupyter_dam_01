## Descarga y almacenaje diario del catalogo en duckdb

```python
import duckdb
import requests
import tempfile
import io
```

```python
url_catalog = "https://datos.madrid.es/egob/catalogo.csv"
path_catalog = '../data/catalogo.csv'
response = requests.get(url=url_catalog)

if response.status_code == 200:
    with open(path_catalog, 'w', encoding='UTF8') as f:
        f.write(response.text)
```

```python
url_catalog = "https://datos.madrid.es/egob/catalogo.csv"
path_catalog = '../data/catalogo.csv'
response = requests.get(url=url_catalog)

if response.status_code == 200:
    with  tempfile.TemporaryFile(mode='w+') as f:
        f.write(response.text)
        f.seek(0)
        rd = duckdb.read_csv(io.StringIO(f.read()), sep=';' , quotechar=None, null_padding=True)

df2 = rd.to_df()
```

```python
url_catalog = "https://datos.madrid.es/egob/catalogo.csv"
path_catalog = '../data/catalogo.csv'
response = requests.get(url=url_catalog)

if response.status_code == 200:
    with  tempfile.TemporaryFile(mode='w+') as f:
        f.write(response.text)
        f.seek(0)
        with  duckdb.connect('../data/duck_catalog.db') as connection:
            rd = connection.read_csv(io.StringIO(f.read()), sep=';' , quotechar=None, null_padding=True)
            rd.to_table('CATALOGO')

```
